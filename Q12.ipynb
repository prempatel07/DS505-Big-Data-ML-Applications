### Problem Statement:
Write a Python program to build and compare two classification models: DecisionTreeClassifier and KNeighborsClassifier (KNN) using Scikit-Learn. Train both models on a classification dataset, evaluate test predictions, and compare their performance using Confusion Matrix and Accuracy Score.

### Expected Reference Output:
```text
--- MODEL EVALUATION & COMPARISON ---
1. Decision Tree Accuracy : 97.78%
Confusion Matrix (Decision Tree):
[[15  0  0]
 [ 0 15  0]
 [ 0  1 14]]

2. K-Nearest Neighbors Accuracy (K=5) : 97.78%
Confusion Matrix (KNN):
[[15  0  0]
 [ 0 15  0]
 [ 0  1 14]]
```

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
````

````python
# Step 1: Load standard Iris multiclass dataset
iris = load_iris()
X, y = iris.data, iris.target
````

````python
# Step 2: Split into Train (70%) and Test (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42, stratify=y)
````

````python
# Step 3: Train Decision Tree Classifier
dt_model = DecisionTreeClassifier(max_depth=3, random_state=42)
dt_model.fit(X_train, y_train)
dt_preds = dt_model.predict(X_test)
dt_acc = accuracy_score(y_test, dt_preds)
````

````python
# Step 4: Train K-Nearest Neighbors Classifier (K=5)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
knn_preds = knn_model.predict(X_test)
knn_acc = accuracy_score(y_test, knn_preds)
````

````python
# Step 5: Compare Performance
print("--- MODEL EVALUATION & COMPARISON ---")
print(f"1. Decision Tree Accuracy : {dt_acc * 100:.2f}%")
print("Confusion Matrix (Decision Tree):")
print(confusion_matrix(y_test, dt_preds))

print(f"\n2. K-Nearest Neighbors Accuracy (K=5) : {knn_acc * 100:.2f}%")
print("Confusion Matrix (KNN):")
print(confusion_matrix(y_test, knn_preds))
````
